```python
# import
import os
import json
from pathlib import Path
import argparse

import cv2
import time
from tqdm import tqdm
from rtmlib import RTMDet, RTMPose

from pose_track import Detection, build_frame_record
```

```python
POC_DIR = Path("/Users/hyeon/SeSAC_Project/PoC")
DETECTOR_ROOT = POC_DIR / 'models' / "detectors" / "rtmdet-nano-person-320x320"
DEVICE = "mps"
```

```python
detector = RTMDet(
    onnx_model=DETECTOR_ROOT/"end2end.onnx",
    model_input_size=(320, 320),
    det_mode="human",
    score_thr=0.4,
    nms_thr=0.45,
    backend="onnxruntime",
    device=DEVICE,
)
```

```python
def detect_image(image_path: Path) -> list[Detection]:
    image = cv2.imread(str(image_path))
    if image is None:
        raise RuntimeError(f"이미지를 읽지 못했습니다: {image_path}")

    det_s = time.perf_counter()
    bboxes = detector(image)
    det_e = time.perf_counter()

    return image, bboxes, det_e - det_s
```
